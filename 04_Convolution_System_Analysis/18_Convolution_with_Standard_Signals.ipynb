# Convolution with Standard Signals (Step & Exponential)

### 🎯 Objective
To analyze the system response when the input signals are standard mathematical sequences:
1.  **Exponential Signal:** $x_1[n] = 2^n u[n]$
2.  **Unit Step Signal:** $x_2[n] = u[n]$ (represented as a sequence of ones)

### 🛠 Description
This notebook extends the previous convolution logic to process longer, mathematically defined sequences.
* **Signal Generation:** We generate sequences for $0 \le n \le 65$.
* **System Response:** We calculate $y_1[n] = x_1[n] * h[n]$ and $y_2[n] = x_2[n] * h[n]$.
* **Visualization:** We plot the inputs and the resulting outputs side-by-side to observe how the system modifies growing (exponential) vs. constant (step) signals.

```python
import numpy as np
import matplotlib.pyplot as plt
def convf(x, n, h, m):
    start = min(n) + min(m)
    N = np.arange(start, start + len(n) + len(m) - 1)
    YY = np.zeros((len(n), len(m)))
    for i in range(len(m)):
        YY[:, i] = x * h[i]
    Y = np.zeros(len(N))
    for NN in range(len(N)):
        for i in range(NN + 1):
            j = NN - i
            if i < len(n) and j < len(m):
                Y[NN] += YY[i, j]
    return Y, N
n1 = np.arange(0, 66)
x1 = 2 ** n1
n2 = np.arange(0, 66)
x2 = np.ones_like(n2)
h = np.array([12, 0, 4, 0, 2, 3, 5])
m = np.arange(-1, -1 + len(h)) 
Y1, Nn1 = convf(x1, n1, h, m)
Y2, Nn2 = convf(x2, n2, h, m)
plt.figure(figsize=(12,8))
plt.suptitle("practise problem-2", fontsize=14, fontweight='bold')
plt.figtext(0.90, 0.96, "ET-231075", ha="right", va="top", fontsize=16, color="k",fontweight='bold')
plt.subplot(3,2,1)
plt.stem(n1, x1)
plt.title("x1(n) = 2^n u(n)")

plt.subplot(3,2,2)
plt.stem(n2, x2)
plt.title("x2(n) = u(n)")

plt.subplot(3,2,3)
plt.stem(m, h)
plt.title("h(n)")

plt.subplot(3,2,4)
plt.stem(m, h)
plt.title("h(n)")

plt.subplot(3,2,5)
plt.stem(Nn1, Y1)
plt.title("y1(n)")

plt.subplot(3,2,6)
plt.stem(Nn2, Y2)
plt.title("y2(n)")

plt.tight_layout()
plt.show()
```
