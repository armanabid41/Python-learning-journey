# Discrete-Time Signal Convolution

### 🎯 Objective
To implement the **Convolution Sum** operation from scratch to calculate the output response of a system.
Formula: $y[n] = x[n] * h[n] = \sum_{k=-\infty}^{\infty} x[k]h[n-k]$

### 🛠 Description
This notebook demonstrates:
1.  **Manual Implementation:** A custom function `convf` is defined to calculate convolution without relying on built-in libraries. This uses the **Tabular Method** (calculating partial products and summing diagonals).
2.  **Signal Generation:** Defining input signal $x[n]$ and impulse response $h[n]$.
3.  **Visualization:** Plotting the Input, Impulse Response, and the resulting Convolved Output to visualize how the system modifies the signal.

```python
import numpy as np
import matplotlib.pyplot as plt

def convf(x, n, h, m):
    start = min(n) + min(m)
    N = np.arange(start, start + len(n) + len(m) - 1)

    YY = np.zeros((len(n), len(m)))
    for i in range(len(m)):
        YY[:, i] = x * h[i]

    Y = np.zeros(len(N))

    for NN in range(len(N)):
        for i in range(NN + 1):
            j = NN - i
            if i < len(n) and j < len(m):
                Y[NN] += YY[i, j]

    return Y, N

x = np.array([2, 3, 1, 0, 7, 5])  
n = np.arange(-4, -4 + len(x))     

h = np.array([12, 0, 5, 0, 2, 3, 5]) 
m = np.arange(-1, -1 + len(h))     

Y, Nn = convf(x, n, h, m)

plt.figure(figsize=(14,10))
plt.suptitle("practise problem-1", fontsize=14, fontweight='bold')
plt.figtext(0.90, 0.96, "ET-231075", ha="right", va="top", fontsize=16, color="k",fontweight='bold')


plt.subplot(3,1,1)
plt.stem(n, x)
plt.title("x(n)")

plt.subplot(3,1,2)
plt.stem(m, h)
plt.title("h(n)")

plt.subplot(3,1,3)
plt.stem(Nn, Y)
plt.title("y(n) = x(n) * h(n)")

plt.tight_layout()
plt.show()
```
