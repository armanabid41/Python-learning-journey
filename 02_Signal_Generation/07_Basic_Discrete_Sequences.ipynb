# Generation of Basic Discrete Time Sequences: Unit Sample, Step, and Ramp

### 🎯 Objective
To implement and plot **fundamental discrete-time sequences** — the unit impulse, unit step, and ramp functions — using NumPy and Matplotlib.

### 🧠 Description
This notebook generates the most common building blocks used in digital signal processing (DSP):  
- **δ[n] (Unit Sample Sequence)**  
- **u[n] (Unit Step Sequence)**  
- **r[n] (Ramp Sequence)**  

Each subplot provides a stem plot representation to visualize how these sequences behave over time.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(14,10))
L=3
M=7
N=5

#subplot-1
t1=np.arange(-(L+M),(L+N)+1)
y1=np.concatenate([np.zeros(abs(-(L+M))),[1],np.zeros(L+N)])

plt.subplot(3,1,1)
plt.stem(t1,y1,linefmt='r-',markerfmt="ko",label='stem')
plt.title('subplot-1:Unit sample sequence')
plt.xlabel('Time(s)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)

#subplot-2
t2=np.arange(-(L+M),(L+N)+1)
y2=np.concatenate([np.zeros(abs(-(L+M))),np.ones(1),np.ones(L+N)])

plt.subplot(3,1,2)
plt.stem(t2,y2,linefmt='g-',markerfmt="ko",label='stem')
plt.title('subplot-2:Unit step sequence')
plt.xlabel('Time(s)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)

#subplot-3
t3 = np.arange(0,(L+M+N))
y3 = t3

plt.subplot(3,1,3)
plt.stem(t3,y3,linefmt='r-',markerfmt="ko",label='stem')
plt.title('subplot-3:Unit Ramp sequence')
plt.xlabel('Time(s)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
