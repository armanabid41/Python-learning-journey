# Plotting Exponential Sequences in Discrete Time Domain

### 🎯 Objective
To generate and visualize **discrete exponential sequences** (both increasing and decaying) using Python and Matplotlib.

### 🧠 Description
This notebook explores the behavior of exponential signals in discrete time.  
It includes multiple examples of exponential growth and decay, shown using stem plots to clearly illustrate the amplitude changes over time.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(14,10))
L=3
M=7
N=5
#subplot-1
n=np.arange(0,(L+M+N)*10)
a1=(M+N)/(M+N+1)
y1=a1**n
plt.subplot(2,2,1)
plt.stem(n,y1,linefmt='r-',markerfmt='ko',label='stem')
plt.title('subplot-1:Exponential sequence')
plt.xlabel("Time(s)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(True)
#subplot-2
n=np.arange(0,(L+M+N)*10)
a2=(M+N+1)/(M+N)
y2=a2**n
plt.subplot(2,2,2)
plt.stem(n,y2,linefmt='g-',markerfmt='ko',label='stem')
plt.title('subplot-2:Exponential sequence')
plt.xlabel("Time(s)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(True)
#subplot-3
n=np.arange(0,(L+M+N)*10)
a3=(-(M+N)/(M+N+5))
y3=a3**n
plt.subplot(2,2,3)
plt.stem(n,y3,linefmt='y-',markerfmt='ro',label='stem')
plt.title('subplot-3:Exponential sequence')
plt.xlabel("Time(s)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(True)
#subplot-4
n=np.arange(0,(L+M+N)*10)
a4=(-(M+N+L)/(M+N))
y4=a4**n
plt.subplot(2,2,4)
plt.stem(n,y4,linefmt='k-',markerfmt='yo',label='stem')
plt.title('subplot-4:Exponential sequence')
plt.xlabel("Time(s)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
