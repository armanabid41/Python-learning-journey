# Time Shifting of Discrete Signals (Delay & Advance)

### 🎯 Objective
To implement and visualize time-shifting operations on discrete-time signals:
1.  **Time Delay:** Shifting the signal to the right ($x[n - k]$).
2.  **Time Advance:** Shifting the signal to the left ($x[n + k]$).

### 🛠 Description
This notebook demonstrates how to:
* Define a custom function `sigshift` to handle index manipulation.
* Create a discrete signal using **NumPy**.
* Apply a **Delay** of $k=9$ units.
* Apply an **Advance** of $k=7$ units.
* Visualize the original, delayed, and advanced signals using `matplotlib.pyplot.stem` to observe the shift in the time axis.

```python
import numpy as np
import matplotlib.pyplot as plt
#Define the signal shift function
def sigshift (x, m, k):
#Implements y(n) = x(n - k)
    n = m + k
    y = x
    return y, n
#Define ID: ET231075
A = 2
L = 3
B = 1
C = 0
M = 5
N = 1
x1 = np.array([2, 3, 1, 1 + 0, 0, 5, 1])
# The arrow is at the 3rd position (B), so n=0 at index 2.
n1 = np.arange(-2, 5) 
# Delay by M+2 positions
#k_delay = M + 2  # k = 7 + 2 = 9
y1, n_delay = sigshift(x1, n1, 9)
# Advance by N+2 positions
#k_advance = -(N + 2)  # k = -(5 + 2) = -7
y2, n_advance = sigshift(x1, n1, -7)
# Plot the signals
plt.figure(figsize=(14, 10))
plt.figtext(0.90, 0.99, "ET-231051", ha="right", va="top", fontsize=16, color="k",fontweight='bold')
# Plot Original Signal
plt.subplot(3, 1, 1)
plt.stem(n1, x1, basefmt="k:",linefmt='r',markerfmt='ko')
plt.title('Signal x1(n1)')
plt.grid(True)
plt.ylabel('Amplitude')
# Plot Delayed Signal
plt.subplot(3, 1, 2)
plt.stem(n_delay, y1, 'r', markerfmt='go', basefmt="k:")
plt.title('Delayed Signal')
plt.grid(True)
plt.ylabel('Amplitude')
# Plot Advanced Signal
plt.subplot(3, 1, 3)
plt.stem(n_advance, y2, 'g', markerfmt='ko', basefmt="k:")
plt.title('Advanced Signal')
plt.grid(True)
plt.xlabel('Time index (n)')
plt.ylabel('Amplitude')
plt.tight_layout()
plt.show()
```
