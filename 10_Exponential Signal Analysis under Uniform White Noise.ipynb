# Exponential Signal with Uniform White Noise

### 🎯 Objective
To study the impact of **uniformly distributed noise** on an **exponentially decaying signal**, and visualize both the noise characteristics and the noisy signal using Matplotlib.

### 🧠 Description
This notebook illustrates:
- Generation of an exponentially decaying signal.  
- Addition of **uniform white noise** to simulate a random disturbance.  
- Visualization of the **noise waveform**, **noise distribution (histogram)**, and the **noisy exponential signal**.

It provides insight into how uniform noise differs from Gaussian noise in amplitude spread and effect on signal smoothness.

```python
import numpy as np
import matplotlib.pyplot as plt
A = 2
L = 3
B = 1
C = 0
M = 7
N = 5

T = 2 / (M + L)            
Fs = 500 * (M + N + L)     
num_samples = int(T * Fs)
t = np.linspace(0, T, num_samples) 

# N=5 is ODD SO DECAYING
signal = A*np.exp(-t)

# N=5 is ODD SO UNIFORM
noise = np.random.uniform(-2, 2, size=len(t))
noisy_signal = signal + noise

plt.figure(figsize=(18, 10))
plt.figtext(0.90, 0.96, "ET-231075", ha="right", va="top", fontsize=16, color="k")

# Plot 1 - Noise
plt.subplot(2, 2, 1)
plt.plot(t, noise, color='y')
plt.title("White Noise (Uniform Distribution)")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")

# Plot 2 - Noise distribution (Histogram)
plt.subplot(2, 2, 2)
plt.hist(noise, bins=50, color='orange', edgecolor='k')
plt.title("Noise Distribution (Uniform)")
plt.xlabel("Noise Amplitude")
plt.ylabel("Frequency")

# Plot 3 - Exponential signal
plt.subplot(2, 2, 3)
plt.plot(t, signal, color='r')
plt.title("Exponential Decaying Signal")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")

# Plot 4 - Noisy signal
plt.subplot(2, 2, 4)
plt.plot(t, noisy_signal, color='g')
plt.title("Noisy Signal (Signal + Uniform White Noise)")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")

plt.show()
```
