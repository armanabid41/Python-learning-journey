# Noise Analysis on Sinusoidal Signal (Gaussian Noise)

### 🎯 Objective
To generate a composite sinusoidal signal and analyze the effect of **additive Gaussian white noise** on it using Python and Matplotlib.

### 🧠 Description
This notebook demonstrates how to:
- Generate a base sinusoidal signal with given parameters.  
- Add Gaussian white noise drawn from a normal distribution.  
- Plot the **original**, **noise**, and **noisy signals**.  
- Display the **noise distribution (PDF)** to verify its Gaussian nature.

Each subplot helps visualize how random noise alters the waveform characteristics in the time domain.


```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# Parameters
M = 7
N = 5
L = 3

T= 2/(M + L)
Fs= 500*(M + N + L)
num_samples = int(T*Fs)
t = np.linspace(0,T,num_samples,endpoint=False)
# Generate Signal
signal =M*np.sin(25*np.pi*(M + L)*t)+N*np.cos(2*np.pi*(N + L)*t)
noise = np.random.normal(0, 1, num_samples)
noisy_signal = signal + noise
plt.figure(figsize=(16, 10))
plt.figtext(0.90, 0.96, "ET-231075", ha="right", va="top", fontsize=16, color="k")

# Plot 1: Noise
plt.subplot(2, 2, 1)
plt.plot(t, noise, color="orange")
plt.title("White Noise (Normal Distribution)")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True)

# Plot 2: Noise Distribution
plt.subplot(2, 2, 2)
# Plot the histogram of the noise
count,bins,ignored=plt.hist(noise,50,density=True,alpha=0.6, color="orange")
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = stats.norm.pdf(x, 0, 1) 
plt.plot(x, p, 'k')
plt.title("Noise Distribution (Normal)")
plt.xlabel("Noise Value")
plt.ylabel("Density")
plt.grid(True)

# Plot 3: Original Signal
plt.subplot(2, 2, 3)
plt.plot(t, signal, color="g")
plt.title("Original Signal")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True)

# Plot 4: Noisy Signal
plt.subplot(2, 2, 4)
plt.plot(t, noisy_signal, color="r")
plt.title("Noisy Signal")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True)

plt.show()
```
