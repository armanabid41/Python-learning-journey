```python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(14,6))
#subplot-1
Fs1=53
T1=1/Fs1
F1=12
A1=3
duration1=19/4
t1 = np.arange(0, duration1+T1, T1)
y1 = A1*np.sin(2*np.pi*F1*t1)
# Line plot
plt.subplot(2,2,1)
plt.plot(t1, y1,"g",linewidth=2, label='continuous plot')
plt.title("subplot-1:continuous plot of sine wave")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(True)

#subplot-2
F2=8       
F3=10      
Fs2=33 
T2=1/Fs2    
duration2=3        
t2 = np.arange(0, duration2+T2, T2)           
y2 = np.sin(2*np.pi*F2*t2) + np.sin(2*np.pi*F3*t2)
# Discrete plot (stem)
plt.subplot(2,2,2)
plt.stem(t2, y2, linefmt="-",markerfmt ='ko', label ='stem ')
plt.title("subplot-2:Discrete Plot")
plt.xlabel("Sample position (t)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(True)

#subplot-3
F4=3
Fs3=29
T3=1/Fs3
duration3=3
A2=3
t3=np.arange(0,duration3+T3,T3)
y3=A2*np.cos(2*np.pi*F4*t3)
#line plot
plt.subplot(2,2,3)
plt.stem(t3,y3,linefmt="--",markerfmt="go",label= 'stem')
plt.plot(t3,y3,linewidth=1,label='plot')
plt.title("subplot-3:Continuous & Discrete Plot of Cosine wave")
plt.xlabel("Sample Position")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(True)

#subplot-4
F5=3
Fs4=29
T4=1/Fs4
duration4=12
A3=3
t4=np.arange(-duration4,duration4+T4,T4)
y4=A3*np.sinc(2*F5*t4)
plt.subplot(2,2,4)
plt.stem(t4,y4,linefmt="--",markerfmt="Dk")
plt.title("subplot-4:Discrete plot of sinc")
plt.xlabel("Sample Position")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(True)


plt.tight_layout()
plt.show()
```
