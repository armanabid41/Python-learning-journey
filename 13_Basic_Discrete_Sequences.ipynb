# Generation of Basic Discrete-Time Sequences

### 🎯 Objective
To generate and visualize fundamental discrete-time signals using Python:
1.  **Unit Sample Sequence:** A custom sequence defined by specific amplitude values.
2.  **Unit Step Sequence:** A signal that is 0 for negative time and 1 for positive time ($u[n]$).
3.  **Unit Ramp Sequence:** A signal where the amplitude increases linearly with time ($r[n] = n$ for $n \ge 0$).

### 🛠 Description
This code demonstrates how to:
* Define specific signal arrays using **NumPy**.
* Generate time indices using `np.arange`.
* Implement logical indexing to create step functions.
* Visualize discrete signals using `matplotlib.pyplot.stem` to create "lollipop" charts appropriate for discrete-time analysis.

```python
import numpy as np
import matplotlib.pyplot as plt
# Define the signal addition function
def sigadd(x1, n1, x2, n2):
    n = n = np. arange (min (min (n1), min (n2)), max (max (n1), max (n2)) + 1)
    # Initialize new signal arrays with zeros
    y1 = np.zeros(len(n))
    y2 = np.zeros(len(n))
    # Place x1 and x2 into their correct time slots in y1 and y2
    y1[(n >= np.min(n1)) & (n <= np.max(n1))] = x1
    y2[(n >= np.min(n2)) & (n <= np.max(n2))] = x2
    y = y1 + y2
    return y, n 
# Define ID: ET231075
A = 2
L = 3
B = 1
C = 0
M = 7
N = 5
# Signal x1(n) and its time index n1
x1 = np.array([2, 3, 1, 1 + 0, 0, 7, 5])
n1 = np.arange(-2, 5)
# Signal x2(n) and its time index n2
x2 = np.array([ 7 + 5, 0, 5 + 0, 0, 2, 3, 1])
n2 = np.arange(-2, 5)
# Add the signals using the sigadd function
y, n = sigadd(x1, n1, x2, n2)
# Plot the signal
plt.figure(figsize=(10, 8))
plt.figtext(0.90, 0.99, "ET-231075")

plt.subplot(3,1,1)
plt.stem(n1,x1,linefmt='r',markerfmt="ko",label='stem')
plt.title('subplot-1:Unit sample sequence')
plt.xlabel('Time(s)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)

#subplot-2
t2=np.arange(-(3+7),(3+5)+1)
y2=np.concatenate([np.zeros(abs(-(3+7))),np.ones(1),np.ones(3+5)])

plt.subplot(3,1,2)
plt.stem(t2,y2,linefmt='g-',markerfmt="ko",label='stem')
plt.title('subplot-2:Unit step sequence')
plt.xlabel('Time(s)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)

#subplot-3
t3 = np.arange(0,15)
y3 = t3

plt.subplot(3,1,3)
plt.stem(t3,y3,linefmt='r-',markerfmt="ko",label='stem')
plt.title('subplot-3:Unit Ramp sequence')
plt.xlabel('Time(s)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
