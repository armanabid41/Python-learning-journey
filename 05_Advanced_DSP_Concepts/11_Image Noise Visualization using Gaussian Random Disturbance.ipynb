# Noise Addition and Visualization on Image

### 🎯 Objective
To analyze and visualize the effect of **additive Gaussian noise** on a grayscale or color image using NumPy and OpenCV.

### 🧠 Description
This notebook demonstrates how to:
- Read and display an image using OpenCV and Matplotlib.  
- Generate Gaussian noise based on image dimensions.  
- Add and clip the noise to maintain valid pixel intensity values.  
- Display the **original image**, **noise image**, and **noisy image** side by side.

This experiment helps understand how noise distorts image data and how pixel-wise random perturbations affect overall visual quality.


```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

image = cv2.imread(r"C:\Users\arman\Downloads\Abid.jpg.png")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

A = 2
L = 3
B = 1
C = 0
M = 7
N = 5

noise = (M + L) + (N * 5) * np.random.randn(*image.shape)
noisy_image = np.clip(image.astype(float) + noise, 0, 255).astype(np.uint8)

plt.figure(figsize=(15, 5))
plt.suptitle("Noise on Image (ID: ET231075)", fontsize=14)

plt.subplot(1, 3, 1)
plt.imshow(image)
plt.title("Original Image by ET231075")
plt.axis('off')

plt.subplot(1, 3, 2)
noise_display = np.clip(noise + 100, 0, 255).astype(np.uint8)
plt.imshow(noise_display, cmap='gray')
plt.title("Noise Image by ET231075")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(noisy_image)
plt.title("Noisy Image by ET231075")
plt.axis('off')

plt.tight_layout()
plt.show()
```
