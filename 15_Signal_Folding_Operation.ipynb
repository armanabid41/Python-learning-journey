# Signal Folding (Time Reversal) Operation

### 🎯 Objective
To implement and visualize the **Folding** (or Reflection) of a discrete-time signal.
Mathematical definition: $y[n] = x[-n]$.

### 🛠 Description
This notebook demonstrates how to:
* Define a custom function `sigfold` to handle the reversal of both signal values and time indices.
* Use **NumPy's** `np.flip()` function to reverse arrays efficiently.
* Visualize the **Original** vs. **Folded** signals using `matplotlib` to observe the mirror image across the vertical axis ($n=0$).

```python
import numpy as np
import matplotlib.pyplot as plt

#Define the signal folding function
def sigfold(x, m):
    #Implements y(n) = x(-n)
    #Time folding (reversal) of a discrete-time signal.

    y = np.flip(x)
    n = -np.flip(m)
    return y, n

#Define ID: ET231075
A = 2
L = 3
B = 1
C = 0
M = 7
N = 5

#Define the original signal x(n)
x1 = np.array([7 + 5, 0, 5 + 0, 0, 2, 3, 5])

# Define its time index 'n'
n1 = np.arange(-2, 5)  # [-2, -1, 0, 1, 2, 3, 4]
#Apply the folding operation
y, n = sigfold(x1, n1)

# Plot the signals
plt.figure(figsize=(14, 10))
plt.figtext(0.90, 0.99, "ET-231075", ha="right", va="top", fontsize=16, color="k",fontweight='bold')

# Plot Original Signal
plt.subplot(2, 1, 1)
plt.stem(n1, x1,markerfmt='go', basefmt="k:")
plt.title('Signal x1(n1)')
plt.grid(True)
plt.ylabel('Amplitude')

# Plot Folded Signal
plt.subplot(2, 1, 2)
plt.stem(n, y, 'r', markerfmt='ro', basefmt="k:")
plt.title('Signal x1(-n1)')
plt.grid(True)
plt.xlabel('Time index (n)')
plt.ylabel('Amplitude')

plt.tight_layout()
plt.show()
```
